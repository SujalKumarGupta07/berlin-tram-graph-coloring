```python
import pandas as pd
import numpy as np

df = pd.read_csv("berlin-tram-2024 (5).csv", header=None)
df = df.fillna(0)
raw = df.to_numpy()
def clean_name(x):
    s = str(x).strip()
    try:
        return str(int(float(s)))
    except ValueError:
        return s
route_names = [clean_name(x) for x in raw[1, 1:]]
conflict = raw[2:, 1:].astype(float)

print("Routes:", route_names)
print("Conflict matrix shape:", conflict.shape)
```

```python
print("Enter the following numbers for paths to get color of that path")
route_index_map = [
    "M1", "M2", "M4", "M5", "M6", "M8", "M10", "12",
    "M13", "16", "M17", "18", "21", "27", "37", "50",
    "60", "61", "62", "63", "67", "68"
]
for idx, name in enumerate(route_index_map):
    print(f"{idx} for {name}")
```

```python
colors1 = [
    "Yellow", "Green", "Blue", "Orange", "Pink", "Grey",
    "Red", "Brown", "Deep Blue", "Purple", "Cyan", "Magenta",
    "Lime", "Teal", "Navy", "Gold", "Maroon", "Olive",
    "Violet", "Turquoise", "Coral", "Indigo"
]
color_groups = [[] for _ in range(len(route_names))]
route_color_map = {}

for x in range(len(route_names)):
    for i, group in enumerate(color_groups):
        if len(group) == 0:
            group.append(x)
            route_color_map[route_names[x]] = colors1[i]
            print("Route: " + route_names[x].ljust(5) + "  Color: " + colors1[i])
            break
        else:
            conflict_found = False
            for j in group:
                if conflict[x, j] == 1 or conflict[j, x] == 1:
                    conflict_found = True
                    break
            if not conflict_found:
                group.append(x)
                route_color_map[route_names[x]] = colors1[i]
                print("Route: " + route_names[x].ljust(5) + "  Color: " + colors1[i])
                break

print()
print("--- Color Groups ---")
for i, group in enumerate(color_groups):
    if group:
        routes_in_group = ", ".join(route_names[r] for r in group)
        print(colors1[i].ljust(12) + ": " + routes_in_group)

total_colors = sum(1 for g in color_groups if g)
print("\nTotal colors used:", total_colors)
```
